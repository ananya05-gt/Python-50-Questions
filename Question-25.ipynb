```python
def calculate_sum(numbers: list[float], index: int = 0) -> float:
    """
    Calculate the sum of list elements using recursion.
    """
    try:
        if not isinstance(numbers, list):
            raise TypeError("Input must be a list.")

        if index == len(numbers):
            return 0

        return numbers[index] + calculate_sum(numbers, index + 1)

    except TypeError as error:
        print("Error:", error)
        return 0


# Example 1
list1 = [1, 2, 3, 4, 5]
print("Sum:", calculate_sum(list1))

# Example 2
list2 = [10, 20, 30]
print("Sum:", calculate_sum(list2))
```
