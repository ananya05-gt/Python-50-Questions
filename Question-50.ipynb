```python
def reverse_list(numbers: list[int]) -> None:
    """
    Reverse the original list in-place using two pointers.

    Args:
        numbers: The list of integers to reverse.

    Returns:
        None. The original list is modified.

    Raises:
        TypeError: If the input is not a list of integers.
    """

    if not isinstance(numbers, list):
        raise TypeError("Input must be a list.")

    if any(not isinstance(num, int) or isinstance(num, bool)
           for num in numbers):
        raise TypeError("All elements must be integers.")

    left = 0
    right = len(numbers) - 1

    while left < right:
        numbers[left], numbers[right] = numbers[right], numbers[left]

        left += 1
        right -= 1


try:
    numbers = [1, 2, 3, 4, 5]

    print("Original list:", numbers)

    reverse_list(numbers)

    print("Reversed list:", numbers)

except TypeError as e:
    print("Error:", e)
```
