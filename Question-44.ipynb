```python
def swap_values(a: int, b: int) -> tuple[int, int]:
    """
    Swap the values of two integers without a temporary variable.

    Args:
        a: First integer.
        b: Second integer.

    Returns:
        A tuple containing the swapped values.

    Raises:
        TypeError: If inputs are not integers.
    """

    if (not isinstance(a, int) or isinstance(a, bool)
            or not isinstance(b, int) or isinstance(b, bool)):
        raise TypeError("Both inputs must be integers.")

    a, b = b, a

    return a, b


try:
    a = 5
    b = 10

    a, b = swap_values(a, b)

    print("After swapping:")
    print("a =", a)
    print("b =", b)

except TypeError as e:
    print("Error:", e)
```
