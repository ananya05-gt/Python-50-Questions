```python
def contains_only_digits(text: str) -> bool:
    """
    Check whether a string contains only numeric digits.

    Args:
        text: The input string.

    Returns:
        True if all characters are digits, otherwise False.

    Raises:
        TypeError: If the input is not a string.
    """

    if not isinstance(text, str):
        raise TypeError("Input must be a string.")

    if text == "":
        return False

    for char in text:
        if char not in "0123456789":
            return False

    return True


try:
    string1 = "123456"
    print(contains_only_digits(string1))

    string2 = "123a56"
    print(contains_only_digits(string2))

except TypeError as e:
    print("Error:", e)
```
