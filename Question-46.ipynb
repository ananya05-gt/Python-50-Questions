```python
def remove_nth_occurrence(text: str, char: str, n: int) -> str:
    """
    Remove the nth occurrence of a character from a string.
    """
    if not isinstance(text, str) or not isinstance(char, str):
        raise TypeError("Text and character must be strings.")

    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n must be an integer.")

    if len(char) != 1:
        raise ValueError("Enter exactly one character.")

    if n < 1:
        raise ValueError("n must be at least 1.")

    count = 0
    result = ""

    for letter in text:
        if letter == char:
            count += 1

            if count == n:
                continue

        result += letter

    if count < n:
        raise ValueError("Occurrence does not exist.")

    return result


try:
    text = "example example example"
    char = 'e'
    n = 3

    print(remove_nth_occurrence(text, char, n))

except (TypeError, ValueError) as e:
    print("Error:", e)
```
