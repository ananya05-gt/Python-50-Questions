```python
def calculate_power(x: float, y: int) -> float:
    """
    Calculate x raised to the power y without using ** operator.
    """
    try:
        if not isinstance(x, (int, float)) or isinstance(x, bool):
            raise TypeError("Base must be a number.")

        if not isinstance(y, int) or isinstance(y, bool):
            raise TypeError("Exponent must be an integer.")

        result = 1

        # Handle negative exponent
        exponent = abs(y)

        for i in range(exponent):
            result = result * x

        if y < 0:
            if x == 0:
                raise ValueError("Zero cannot have a negative exponent.")
            result = 1 / result

        return result

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return 0


# Main program
try:
    base = float(input("Enter the base (x): "))
    exponent = int(input("Enter the exponent (y): "))

    result = calculate_power(base, exponent)

    print("Result:", result)

except ValueError:
    print("Error: Please enter valid numeric values.")
```
