```python
def find_common_elements(
    list1: list[int],
    list2: list[int],
    list3: list[int]
) -> list[int]:
    """
    Find unique elements common to all three lists.

    Args:
        list1: First list of integers.
        list2: Second list of integers.
        list3: Third list of integers.

    Returns:
        A list of unique common elements in list1 order.

    Raises:
        TypeError: If inputs are not lists of integers.
    """

    for numbers in (list1, list2, list3):
        if not isinstance(numbers, list):
            raise TypeError("All inputs must be lists.")

        if any(not isinstance(num, int) or isinstance(num, bool)
               for num in numbers):
            raise TypeError("All list elements must be integers.")

    common = []

    for num in list1:
        if num in list2 and num in list3 and num not in common:
            common.append(num)

    return common


try:
    list1 = [1, 2, 3, 78, 4, 5, 5, 2, 3]
    list2 = [3, 4, 5, 6, 7, 78]
    list3 = [5, 6, 7, 8, 9, 78]

    result = find_common_elements(list1, list2, list3)

    print("Common elements:", result)

except TypeError as e:
    print("Error:", e)
```
