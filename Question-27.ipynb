```python
def merge_sorted_lists(list1: list[int], list2: list[int]) -> list[int]:
    """
    Merge two sorted lists into one sorted list.
    """
    try:
        if not isinstance(list1, list) or not isinstance(list2, list):
            raise TypeError("Both inputs must be lists.")

        if not all(type(num) is int for num in list1 + list2):
            raise ValueError("All elements must be integers.")

        if list1 != sorted(list1) or list2 != sorted(list2):
            raise ValueError("Both lists must already be sorted.")

        result = []
        i = 0
        j = 0

        # Compare elements from both lists
        while i < len(list1) and j < len(list2):

            if list1[i] <= list2[j]:
                result.append(list1[i])
                i += 1
            else:
                result.append(list2[j])
                j += 1

        # Add remaining elements from list1
        while i < len(list1):
            result.append(list1[i])
            i += 1

        # Add remaining elements from list2
        while j < len(list2):
            result.append(list2[j])
            j += 1

        return result

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return []


# Example 1
list1 = [1, 3, 5]
list2 = [2, 4, 6]

print("Merged list:", merge_sorted_lists(list1, list2))

# Example 2
list3 = [10, 20, 30]
list4 = [15, 25, 35]

print("Merged list:", merge_sorted_lists(list3, list4))
```
