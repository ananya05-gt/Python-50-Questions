```python
def is_power_of_two(number: int) -> bool:
    """
    Check whether a number is a power of 2.
    """
    try:
        if type(number) is not int:
            raise TypeError("Input must be an integer.")

        if number <= 0:
            return False

        while number % 2 == 0:
            number = number // 2

        return number == 1

    except TypeError as error:
        print("Error:", error)
        return False


# Example 1
number1 = 16
print("Is power of 2:", is_power_of_two(number1))

# Example 2
number2 = 18
print("Is power of 2:", is_power_of_two(number2))
```
