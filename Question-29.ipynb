```python
def count_word_frequency(filename: str) -> dict[str, int]:
    """
    Read a text file and count the frequency of each word.
    """
    try:
        word_count = {}

        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()

        words = content.split()

        for word in words:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1

        return word_count

    except FileNotFoundError:
        print("Error: File not found.")
        return {}

    except OSError as error:
        print("File error:", error)
        return {}


# Call the function
filename = "sample.txt"

result = count_word_frequency(filename)

# Display word frequency
for word, frequency in result.items():
    print(f"{word}: {frequency}")
```
