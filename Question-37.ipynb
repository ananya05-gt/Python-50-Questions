```python
def find_pairs(numbers: list[int], target: int) -> list[tuple[int, int]]:
    """
    Find all unique pairs whose sum equals the target number.
    """
    try:
        if not isinstance(numbers, list):
            raise TypeError("Input must be a list.")

        if not all(type(num) is int for num in numbers):
            raise ValueError("All list elements must be integers.")

        if type(target) is not int:
            raise TypeError("Target must be an integer.")

        pairs = []

        for i in range(len(numbers)):
            for j in range(i + 1, len(numbers)):

                if numbers[i] + numbers[j] == target:
                    pairs.append((numbers[i], numbers[j]))

        return pairs

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return []


# Example
numbers = [1, 2, 3, 4, 5, 6]
target = 7

result = find_pairs(numbers, target)

print("Pairs:", result)
```
