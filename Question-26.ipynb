```python
def sort_list(numbers: list[int]) -> list[int]:
    """
    Sort a list of integers in ascending order using Bubble Sort.
    """
    try:
        if not isinstance(numbers, list):
            raise TypeError("Input must be a list.")

        if not all(type(num) is int for num in numbers):
            raise ValueError("All list elements must be integers.")

        result = numbers.copy()
        n = len(result)

        for i in range(n):
            for j in range(0, n - i - 1):

                if result[j] > result[j + 1]:
                    result[j], result[j + 1] = result[j + 1], result[j]

        return result

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return []


# Example
numbers = [45, 20, 3, 40, 5, 20]

result = sort_list(numbers)

print("Sorted list:", result)
```
