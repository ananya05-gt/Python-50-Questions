```python
def find_second_largest(numbers: list[int]) -> int | None:
    """
    Find the second largest distinct element in a list.

    Returns None if a second distinct largest element does not exist.
    """
    try:
        if not isinstance(numbers, list):
            raise TypeError("Input must be a list.")

        if not numbers:
            raise ValueError("The list cannot be empty.")

        largest = None
        second_largest = None

        for num in numbers:
            if not isinstance(num, int) or isinstance(num, bool):
                raise TypeError("All list elements must be integers.")

            if largest is None or num > largest:
                second_largest = largest
                largest = num

            elif num != largest and (
                second_largest is None or num > second_largest
            ):
                second_largest = num

        return second_largest

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return None


# Main program
try:
    my_list = [7, 5, 8, 2, 10, 9]

    result = find_second_largest(my_list)

    print("Original list:", my_list)
    print("Second largest element:", result)

except Exception as error:
    print("Unexpected error:", error)
```
