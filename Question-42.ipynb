```python
def find_lcm(num1: int, num2: int) -> int:
    """
    Calculate the LCM of two integers.

    Args:
        num1: First integer.
        num2: Second integer.

    Returns:
        The least common multiple of the two numbers.

    Raises:
        TypeError: If inputs are not integers.
    """

    if (not isinstance(num1, int) or isinstance(num1, bool)
            or not isinstance(num2, int) or isinstance(num2, bool)):
        raise TypeError("Both inputs must be integers.")

    if num1 == 0 or num2 == 0:
        return 0

    a = abs(num1)
    b = abs(num2)

    # Find GCD using the Euclidean algorithm
    x, y = a, b

    while y != 0:
        x, y = y, x % y

    gcd = x

    # Calculate LCM
    lcm = (a * b) // gcd

    return lcm


try:
    num1 = 24
    num2 = 9

    result = find_lcm(num1, num2)

    print("LCM is:", result)

except TypeError as e:
    print("Error:", e)
```
