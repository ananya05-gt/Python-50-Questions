```python
def decimal_to_binary(decimal: int) -> str:
    """
    Convert a non-negative decimal integer into binary
    without using the bin() function.
    """
    try:
        if not isinstance(decimal, int) or isinstance(decimal, bool):
            raise TypeError("Input must be an integer.")

        if decimal < 0:
            raise ValueError("Enter a non-negative integer.")

        if decimal == 0:
            return "0"

        binary = ""

        while decimal > 0:
            remainder = decimal % 2
            binary = str(remainder) + binary
            decimal = decimal // 2

        return binary

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return ""


# Main program
try:
    number = int(input("Enter a decimal number: "))

    result = decimal_to_binary(number)

    print("Binary representation:", result)

except ValueError:
    print("Error: Please enter a valid integer.")
```
