```python
def first_non_repeated(text: str) -> str | None:
    """
    Find the first character that does not repeat in a string.

    Args:
        text: The input string.

    Returns:
        The first non-repeated character or None if none exists.

    Raises:
        TypeError: If the input is not a string.
    """

    if not isinstance(text, str):
        raise TypeError("Input must be a string.")

    for char in text:
        if text.count(char) == 1:
            return char

    return None


try:
    text = "swiss"
    result = first_non_repeated(text)

    print("First non-repeated character:", result)

    text = "repeated"
    result = first_non_repeated(text)

    print("First non-repeated character:", result)

except TypeError as e:
    print("Error:", e)
```
