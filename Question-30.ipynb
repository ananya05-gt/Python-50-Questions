```python
def find_missing_number(numbers: list[int]) -> int:
    """
    Find the missing number in a consecutive sequence.
    """
    try:
        if not isinstance(numbers, list):
            raise TypeError("Input must be a list.")

        if len(numbers) < 2:
            raise ValueError("List must contain at least two numbers.")

        if not all(type(num) is int for num in numbers):
            raise ValueError("All elements must be integers.")

        if numbers != sorted(numbers):
            raise ValueError("List must be in ascending order.")

        if len(set(numbers)) != len(numbers):
            raise ValueError("List must not contain duplicate numbers.")

        for i in range(len(numbers) - 1):
            if numbers[i + 1] - numbers[i] > 1:
                return numbers[i] + 1

        raise ValueError("No missing number found.")

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return -1


# Example 1
list1 = [1, 2, 3, 5, 6]
print("Missing number:", find_missing_number(list1))

# Example 2
list2 = [10, 11, 12, 14, 15]
print("Missing number:", find_missing_number(list2))
```
