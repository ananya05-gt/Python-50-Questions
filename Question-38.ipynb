```python
def tuples_to_dictionary(data: list[tuple[str, int]]) -> dict[str, int]:
    """
    Convert a list of key-value tuples into a dictionary.
    """
    try:
        if not isinstance(data, list):
            raise TypeError("Input must be a list.")

        result = {}

        for item in data:
            if not isinstance(item, tuple) or len(item) != 2:
                raise ValueError("Each item must be a tuple containing two elements.")

            key, value = item

            if not isinstance(key, str) or type(value) is not int:
                raise ValueError("Each tuple must contain a string key and an integer value.")

            result[key] = value

        return result

    except (TypeError, ValueError) as error:
        print("Error:", error)
        return {}


# Example
data = [('a', 1), ('b', 2), ('c', 3), ('d', 4)]

result = tuples_to_dictionary(data)

print("Dictionary:", result)
```
