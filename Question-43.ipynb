```python
def binary_to_decimal(binary: str) -> int:
    """
    Convert a binary string into a decimal integer.

    Args:
        binary: A string containing only 0s and 1s.

    Returns:
        The decimal equivalent of the binary number.

    Raises:
        TypeError: If the input is not a string.
        ValueError: If the string is empty or contains invalid characters.
    """

    if not isinstance(binary, str):
        raise TypeError("Input must be a string.")

    if not binary:
        raise ValueError("Binary string cannot be empty.")

    decimal = 0

    for digit in binary:
        if digit not in "01":
            raise ValueError("Binary number can contain only 0 and 1.")

        decimal = decimal * 2 + int(digit)

    return decimal


try:
    binary = "1011"

    result = binary_to_decimal(binary)

    print("Decimal number:", result)

except (TypeError, ValueError) as e:
    print("Error:", e)
```
